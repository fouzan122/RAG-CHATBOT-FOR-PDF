```python
import os
#from langchain.document_loaders import UnstructuredFileLoader #old import not working
from langchain_community.document_loaders import UnstructuredFileLoader
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_groq import ChatGroq
from langchain_classic.chains import RetrievalQA
```

```python
os.environ["GROQ_API_KEY"] = "GROQ_API_KEY"
```

```python
import requests
url = "https://dspmuranchi.ac.in/pdf/Blog/Python%20Built-In%20Functions.pdf"
response = requests.get(url)
```

```python
with open("python_inbuildfunction.pdf", "wb") as f:
    f.write(response.content)
```

```python
loader = UnstructuredFileLoader("python_inbuildfunction.pdf")
```

```python
documents = loader.load()
```

```python
!pip install "unstructured[pdf]"
```

```python
text_splitter = CharacterTextSplitter(
    chunk_size=2000,
    chunk_overlap=400
    )
```

```python
texts = text_splitter.split_documents(documents)
```

```python
type(texts)
```

```python
len(texts)
```

```python
texts[1]
```

```python
embeddings = HuggingFaceEmbeddings()
```

```python
persist_directory = "vector_db"
```

```python
vectordb = Chroma.from_documents(
    documents=texts,
    embedding=embeddings,
    persist_directory=persist_directory
)
```

```python
#retriever
retriever = vectordb.as_retriever()
```

```python
#llm from groq
llm = ChatGroq(
    model_name="qwen/qwen3.8-27b",
    temperature=0
)
```

```python
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=retriever,
    return_source_documents=True
)
```

```python
query = "what are the function from this pdf"
response = qa_chain.invoke({"query": query})
```

```python
import gradio as gr

def rag_chat(message, history):
    try:
        response = qa_chain.invoke({
            "query": message
        })

        answer = response["result"]

        source_documents = response.get("source_documents", [])

        sources = []

        for doc in source_documents:
            source = doc.metadata.get("source", "unknown")
            page = doc.metadata.get("page", "N/A")

            source_info = f"{source} - page: {page}"

            if source_info not in sources:
                sources.append(source_info)

        if sources:
            answer += "\n\n### Sources\n"

            for source_info in sources:
                answer += f"- {source_info}\n"

        return answer

    except Exception as e:
        return f"Error: {str(e)}"


demo = gr.ChatInterface(
    fn=rag_chat,
    title="RAG Chatbot for PDF",
    description="Ask questions about your PDF document.",
    examples=[
        "Give me a summary of this PDF.",
        "What are the main topics in this document?",
        "Explain the important concepts."
    ]
)

if __name__ == "__main__":
    demo.launch()
```
